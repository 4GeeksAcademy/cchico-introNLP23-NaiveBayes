```python
import pandas as pd
```

```python
df = pd.read_csv("https://breathecode.herokuapp.com/asset/internal-link?id=932&path=url_spam.csv")
df.head()
```

```python
import re

def limpiar_texto(texto):
    texto = texto.lower()                         # minúsculas
    texto = re.sub(r'[^a-zA-Záéíóúñü]', ' ', texto)  # reemplazar símbolos por espacio
    texto = re.sub(r'\s+', ' ', texto).strip()    # eliminar espacios duplicados
    return texto

df["url_limpio"] = df["url"].apply(limpiar_texto)

print(df["url_limpio"].head())
```

```python
df["Comentario"] = df["url_limpio"]
```

```python
etiquetas = df["is_spam"]
df["Resultado Esperado"] = etiquetas
```

```python
df_model = df[["url_limpio"]].rename(columns={"url_limpio": "Comentario"})
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df["url_limpio"])

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
vect_df = pd.DataFrame(X.todense(), columns=vectorizer.get_feature_names_out())
vect_df["Resultado Esperado"] = ["Positivo","Negativo","Negativo","Positivo"]
vect_df
```
